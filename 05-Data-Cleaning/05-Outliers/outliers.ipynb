```python
import pandas as pd

age = [19, 20, 21, 22, 23, 250]
df = pd.DataFrame({"age": age})

print(df["age"].median()) # Output: 21.5
print(df["age"].mean())   # Output: 59.166666666666664
```

```python
q1 = df["age"].quantile(0.25)
q3 = df["age"].quantile(0.75)
iqr = q3 - q1

low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr
print("Q1, Q3, IQR:", q1, q3, iqr)
print("Limits:", low, high)

# Using lowercase "age" to match your dataframe
print(df[(df["age"] < low) | (df["age"] > high)])
```

```python
#Clapping
marks = [45, 55, 85, 95, 60, 100] # Example data
df["marks"] = marks
print(df["marks"].clip(50, 90).tolist())
```

```python
#Create A DataFrame with names in mixed case and extra spaces. clean them using strip() and title().
import pandas as pd

# Create a DataFrame with messy names
data = {"names": ["   KrUshNA  ", " SaChIn   ", "gaUrAv   "]}
df_names = pd.DataFrame(data)

print("Original:")
print(df_names)
print("-" * 20)

# Clean the names using the .str accessor
df_names["names"] = df_names["names"].str.strip().str.title()

print("Cleaned:")
print(df_names)
```

```python
#Convert a column of the numbers stored as text to integres. handles a bad values with to_numeric(errors="coerce").
import pandas as pd
import numpy as np

# Create a DataFrame with numbers as text and a bad value
df_nums = pd.DataFrame({"text_numbers": ["10", "25", "error_value", "42", "99"]})

print("Original Data:")
print(df_nums)
print("\nOriginal Dtype:", df_nums["text_numbers"].dtype)
print("-" * 30)

df_nums["text_numbers"] = pd.to_numeric(df_nums["text_numbers"], errors="coerce")

df_nums["text_numbers"] = df_nums["text_numbers"].astype("Int64")

print("\nConverted Data:")
print(df_nums)
print("\nNew Dtype:", df_nums["text_numbers"].dtype)
```

```python
#Add duplicates rows to a DataFrame, Count them & remove them.
import pandas as pd

# 1. Create a DataFrame with duplicate rows
data = {
    "ID": [1, 2, 3, 2, 4, 5, 4],
    "Name": ["Alice", "Bob", "Charlie", "Bob", "David", "Eve", "David"]
}
df_dup = pd.DataFrame(data)

print("Original DataFrame:")
print(df_dup)
print("-" * 30)

# 2. Count the duplicate rows
duplicate_count = df_dup.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
print("-" * 30)

# 3. Remove the duplicate rows
# use inplace=True to modify the original dataframe, or reassign it
df_clean = df_dup.drop_duplicates()

print("DataFrame after removing duplicates:")
print(df_clean)
```

```python
import pandas as pd

data = {"city": ["New York", "Paris", "London", "Paris", "Tokyo", "London"]}
df_cities = pd.DataFrame(data)

print("Original DataFrame:")
print(df_cities)
print("-" * 30)

enc = pd.get_dummies(df_cities[["city"]], dtype=int)

print("One-Hot Encoded DataFrame:")
print(enc)

print("-" * 30)
print("Transposed (as in your original code):")
print(enc.T)
```
