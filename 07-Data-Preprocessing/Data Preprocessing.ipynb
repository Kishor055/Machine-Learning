```python
import pandas as pd
import numpy as np

raw = pd.DataFrame({
    "name" : ["Riya", "aman", "Neha", "Karan",
              "aman", "Isha", "Rohan"],
    "age" : [21, 22, np.nan, 23, 22, 20, 250],
    "city" :["Pune", "mumbai", "pune", None,
             "mumbai", "kolhapur", "PUNE"],
    "marks" :["85", "62", "91", "45",
              "62", "78", "70"],
    "grade" : ["A", "B", "A", "C" ,"B" ,"B", "B"]
})
print(raw)
raw.info()
```

```python
#Always Work On A copy
df = raw.copy()#keep the original safe
#String Clean With .str
df["name"] = df["name"].str.strip().str.title()
df["city"] = df["city"].str.title()

print(df["name"].head(3).tolist())
print(df["city"].value_counts())
```

```python
new = df["city"].replace({"Kolhapur" : "Kop"})
print(new.value_counts().to_dict())
```

```python
#Fixing Data Types 
df["marks"] = df["marks"].astype(int)
print(df.dtypes)
```

```python
#when conversion fails
#astype(int) crashes if one value is not a number.to_numeric() with errors="coerce" turns bad values into NaN instead.
s = pd.Series(["10", "x", "30"])
print(pd.to_numeric(s, errors="coerce").tolist())
```

```python
#dates
dates = ["2026-01-15", "2026-02-20"]
d = pd.to_datetime(pd.Series(dates))

print(d.dt.month.tolist())
print(d.dt.day_name().tolist())
```

```python
#find Duplicates and Missing Value
#Remove the repeates and fill the gaps
print("Duplicates:", df.duplicated().sum())
print(df[df.duplicated(keep=False)])
```

```python
#duplicates removing
df = df.drop_duplicates().reset_index(drop=True)
print(df)
```

```python
#Finding Missing Value
print(df.isnull().sum())
```
